```python
import statistics
import pandas as pd
```

```python
url = (
    "https://raw.githubusercontent.com/"
    "natacherchum/Descriptive_Statistics/"
    "refs/heads/main/descriptive_statistics_example1.csv"
)

df = pd.read_csv(url)

display(df)
```

> **Important!** Functions in the `statistics` module do not have `skipna` or `nan_policy` parameters, so missing values must be removed **before calculations**.

```python
results = {}

for name in df.columns:
    # Remove missing values and convert the column to a Python list.
    x = df[name].dropna().tolist()

    # Calculate quartiles using the inclusive method.
    q1, median, q3 = statistics.quantiles(
        x,
        n=4,
        method="inclusive",
    )

    results[name] = {
        "Available observations": len(x),
        "Missing values": int(df[name].isna().sum()),
        "Mean": statistics.mean(x),
        "Median": statistics.median(x),
        "Minimum": min(x),
        "Maximum": max(x),
        "Range": max(x) - min(x),
        "Q1": q1,
        "Q3": q3,
        "IQR": q3 - q1,
        "Population variance": statistics.pvariance(x),
        "Sample variance": statistics.variance(x),
        "Population standard deviation": statistics.pstdev(x),
        "Sample standard deviation": statistics.stdev(x),
    }

summary = pd.DataFrame(results)

display(summary)
```

### Conclusion

The `statistics` module works with one-dimensional sequences.
Each variable is therefore processed separately.

Missing values are removed before calculations.

Separate functions are provided for population and sample variance
and standard deviation.

Quartiles use `method="inclusive"` to match the linear method
used in the earlier examples. The default method in
`statistics.quantiles` is `"exclusive"`.

Skewness and kurtosis are not provided by this module.
