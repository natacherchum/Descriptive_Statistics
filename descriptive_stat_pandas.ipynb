```python
import pandas as pd
```

```python
url = (
    "https://raw.githubusercontent.com/"
    "natacherchum/Descriptive_Statistics/"
    "refs/heads/main/descriptive_statistics_example1.csv"
)

df = pd.read_csv(url)

display(df)
print("Shape:", df.shape)
```

### Missing Values in pandas

By default, `mean()` ignores missing values (`skipna=True`).

With `skipna=False`, the result is NaN for any column
containing a missing value.

Missing values are excluded separately for each column.
They are not replaced with zeros.

```python
display(df.describe())
```

### Interpretation

`df.describe()` summarizes each numerical variable:

- `count` — number of non-missing observations.
- `mean` — arithmetic mean.
- `std` — sample standard deviation (`ddof=1`).
- `min` and `max` — minimum and maximum.
- `25%`, `50%`, and `75%` — quartiles; `50%` is the median.

Missing values are excluded separately for each column.
Height has 20 observations, while weight and activity have 19 each.

Variance, mode, skewness, and kurtosis are not included.

```python
summary = pd.DataFrame(
    {
        "Mean": df.mean(),
        "Median": df.median(),
        "Minimum": df.min(),
        "Maximum": df.max(),
        "Range": df.max() - df.min(),
        "Q1": df.quantile(0.25),
        "Q3": df.quantile(0.75),
        "IQR": df.quantile(0.75) - df.quantile(0.25),
        "Population variance": df.var(ddof=0),
        "Sample variance": df.var(ddof=1),
        "Population standard deviation": df.std(ddof=0),
        "Sample standard deviation": df.std(ddof=1),
        "Skewness": df.skew(),
        "Excess kurtosis": df.kurt(),
    }
)

display(summary)
```
