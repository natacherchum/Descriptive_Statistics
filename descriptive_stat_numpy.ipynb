```python
import numpy as np
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/natacherchum/Descriptive_Statistics/refs/heads/main/descriptive_statistics_example1.csv"

values = np.genfromtxt(
    url,
    delimiter=",",
    skip_header=1,
    dtype=float,
    filling_values=np.nan,
)

print("Object type:", type(values))
print("Shape:", values.shape)
print("Number of dimensions:", values.ndim)
print("Data type:", values.dtype)
```

### Conclusion

All three variables are stored in a single two-dimensional NumPy array.

- The array contains **20 rows and 3 columns**.
- Rows represent participants.
- Columns represent height, weight, and daily physical activity.
- Every element has the same data type: **float64**.

Here, homogeneous means that all elements share the same data type,
even though the variables have different units of measurement.

```python
variable_names = ["height_cm", "weight_kg", "activity_minutes"]

summary = pd.DataFrame(
    {
        "Missing values": np.isnan(values).sum(axis=0),
        "Mean: np.mean": np.mean(values, axis=0),
        "Mean: np.nanmean": np.nanmean(values, axis=0),
    },
    index=variable_names,
)

display(summary)
```

### Conclusion

`np.mean` returns NaN for a column containing a missing value.

`np.nanmean` ignores missing values and calculates the mean
using only the available observations.

Missing values are not replaced with zeros.

NumPy provides similar functions for other statistics:

- `np.nanmedian` — median.
- `np.nanmin` and `np.nanmax` — minimum and maximum.
- `np.nanvar` — variance.
- `np.nanstd` — standard deviation.
- `np.nanquantile` — quantiles.

These functions also ignore NaN values when performing calculations.

```python
results = {}

for column, name in enumerate(variable_names):
    x = values[:, column]

    q1, median, q3 = np.nanquantile(x, [0.25, 0.50, 0.75])

    results[name] = {
        "Valid observations": np.count_nonzero(~np.isnan(x)),
        "Missing observations": np.count_nonzero(np.isnan(x)),
        "Mean": np.nanmean(x),
        "Median": median,
        "Minimum": np.nanmin(x),
        "Maximum": np.nanmax(x),
        "Range": np.nanmax(x) - np.nanmin(x),
        "Q1": q1,
        "Q3": q3,
        "Interquartile range": q3 - q1,
        "Population variance": np.nanvar(x, ddof=0),
        "Sample variance": np.nanvar(x, ddof=1),
        "Population standard deviation": np.nanstd(x, ddof=0),
        "Sample standard deviation": np.nanstd(x, ddof=1),
    }

summary = pd.DataFrame(results)

# Display two decimal places without changing the stored values.
display(summary)
```
